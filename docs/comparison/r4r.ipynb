# R4R

R4R derives its API from a directory tree and passes results through Apache Velocity templates the developer writes. It binds a single endpoint for the whole server, so it answers from OpenCitations Meta only, and labels every response `application/json`.

```python
from helper import call
```

## A simple request

Looking up a DOI returns the article's OMID and title from OpenCitations Meta.

```python
call("http://localhost:8083/articles?doi=10.1007/s11192-022-04367-w")
```

## The join

No join. A single endpoint is bound for the whole server, so R4R cannot reach OpenCitations Index to add the reference count.

## Output

Whatever the template writes.

```python
call("http://localhost:8083/articles/061202127149")
```

## Pagination

`size` and `offset` page the result, where `offset` is a page index; no `Link` headers and no total. The agents are a nested resource (a subfolder).

```python
call("http://localhost:8083/articles/061202127149/agents?size=2&offset=1")
```

## Versioning

Not supported.

## API description

Not supported: `/doc` serves only static files placed there by hand.

## Consumer authentication

Global: `API_USERS` guards every operation, not only writes. The `r4r-auth` service (port 8084) is the same image with `API_USERS=demo:demo`. Without credentials the request is rejected with 401.

```python
call("http://localhost:8084/articles/061202127149")
```

With the credentials the same request succeeds.

```python
call("http://localhost:8084/articles/061202127149", basic_auth=("demo", "demo"))
```

## Endpoint authentication

Not supported.

## Operations

`GET` only. The other methods are rejected.

```python
call("http://localhost:8083/articles/061202127149", method="POST", max_lines=0)
call("http://localhost:8083/articles/061202127149", method="PUT", max_lines=0)
call("http://localhost:8083/articles/061202127149", method="DELETE", max_lines=0)
```

## Caching

Not supported. The tool reaches OpenCitations Meta through a proxy that records every request it forwards, and two identical calls send the same SPARQL requests twice.

```python
from helper import count_endpoint_requests

count_endpoint_requests("http://localhost:8083/articles?doi=10.1007/s11192-022-04367-w")
```
